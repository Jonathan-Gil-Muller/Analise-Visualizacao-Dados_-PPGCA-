## Listas

```python
# usar [ ] para definir uma lista
lista_notas = [6.0, 7.0, 8.0]

# A função 'sum' soma todos os valores da lista
# A função 'len' retorna o total de elementos da lista
media = sum(lista_notas) / len(lista_notas)

print(media)
```

## Dicionários, dataframes e bibliotecas

```python
# Usar { } para definir um dicionário
# {'nome 01': valor 01, 'nome 02': valor 02, ...}
dicionario = {
    'nomes':['nomeA', 'nomeB', 'nomeC'],
    'nota1':[5.0, 7.0, 8.0],
    'nota2':[5.0, 5.6, 9.8],
    'nota3':[7.0, 4.6, 7.8]
}

print(dicionario)
```

```python
# Uma biblioteca é como uma 'caixa de ferramentas' com várias funções já programadas
# 'pandas' é o nome da biblioteca e 'pd' é o apelido.
# Pandas é uma biblioteca com funções para manipulação de tabelas
import pandas as pd

df = pd.DataFrame(dicionario)

df.head()
```

```python
# Numpy é uma biblioteca para cálculos matemáticos
import numpy as np

media_n1 = np.mean(df['nota1'])

print(media_n1)
```

```python
# Criar uma nova coluna com a média dos alunos

df['Média'] = ((df['nota1'] + df['nota2'] + df['nota3']) / 3).round(1)

df.head()
```
