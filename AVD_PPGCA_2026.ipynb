```python
# Importando a função de transformação ordinal do scikit-learn
from sklearn.preprocessing import OrdinalEncoder

ordem = ['Primeiro', 'Segundo', 'Terceiro', 'Quarto', 'Quinto']
codificador = OrdinalEncoder(categories=[ordem])  # atribui valores numéricos baseados na ordem da lista
df['Clas_convertida'] = codificador.fit_transform(df[['Classificação']])

print(df)
```

```python
######################################
## CONVERSÃO: NOMINAL PARA NUMÉRICO ##
######################################

from sklearn.preprocessing import OneHotEncoder

# criar uma coluna binária para cada categoria única encontrada
encoder = OneHotEncoder()

# transformar a coluna categórica em uma matriz binária (Dummificação)
estados = encoder.fit_transform(df[['Estado']]).toarray()

# converte a matriz de números binários em um DataFrame do Pandas
estados_df = pd.DataFrame(estados, columns=encoder.categories_[0])

# Concatenando as novas colunas e removendo a coluna original
df = pd.concat([df, estados_df], axis=1)
df.drop('Estado', axis=1, inplace=True)

print(df)
```

**Normalização**

```python
import pandas as pd

dados = {
    'Candidato': ['Ana', 'Bruno', 'Carlos', 'Daniela', 'Eduardo'],
    'Anos_Experiencia': [1, 3, 5, 8, 10],
    'Pontuacao_Teste': [400, 550, 700, 850, 1000]
}
df = pd.DataFrame(dados)
df.head()
```

```python
# Aplicar as transformações na coluna 'Pontuacao_Teste'
x = df['Pontuacao_Teste']
min_x = x.min()
max_x = x.max()
media_x = x.mean()
desvio_x = x.std()

# 1. Normalização pelo Máximo
df['Teste_Max'] = x / max_x

# 2. Normalização Min-Max ([0, 1])
df['Teste_MinMax'] = (x - min_x) / (max_x - min_x)

# 3. Escalonamento em Intervalo Personalizado [a, b]
a, b = -2, 2
df['Teste_Intervalo'] = a + ((x - min_x) * (b - a)) / (max_x - min_x)

# 4. Padronização Z-Score (Média = 0, Desvio Padrão = 1)
df['Teste_ZScore'] = (x - media_x) / desvio_x


print(df[['Candidato', 'Pontuacao_Teste', 'Teste_Max', 'Teste_MinMax',
          'Teste_Intervalo', 'Teste_ZScore']].round(3))
```

```python
import numpy as np
from sklearn.preprocessing import RobustScaler, StandardScaler, MinMaxScaler

dados = {
    'Funcionario': ['Ana', 'Bruno', 'Carlos', 'Daniela', 'Eduardo', 'Fernanda',
                    'Gabriel', 'Helena', 'Pedro', 'Alberto'],
    'Horas_Estudo': [2, 3, 3, 4, 4, 5, 5, 6, 45, 60],
    'Salario_k': [3.5, 4.0, 4.2, 4.5, 5.0, 5.2, 5.5, 6.0, 85.0, 51.0]
}

df = pd.DataFrame(dados)

# Instanciando os escaladores para comparação
robust = RobustScaler()
standard = StandardScaler()
minmax = MinMaxScaler()

# Aplicando as transformações na coluna 'Salario_k'
coluna = df[['Salario_k']]

df['Salario_Robust'] = robust.fit_transform(coluna).round(2)
df['Salario_Standard'] = standard.fit_transform(coluna).round(2)
df['Salario_MinMax'] = minmax.fit_transform(coluna).round(2)

print(df[['Funcionario', 'Salario_k', 'Salario_Robust', 'Salario_Standard',
          'Salario_MinMax']])
```

```python
# Aplicando primeiro a transformação logarítmica
df['Salario_Log'] = np.log(df['Salario_k'])

# Instanciando os escaladores
minmax = MinMaxScaler()
standard = StandardScaler()
robust = RobustScaler()

# Aplicando os escaladores sobre a coluna transformada em log
df['Log_MinMax'] = minmax.fit_transform(df[['Salario_Log']]).round(2)
df['Log_Standard'] = standard.fit_transform(df[['Salario_Log']]).round(2)
df['Log_Robust'] = robust.fit_transform(df[['Salario_Log']]).round(2)

# Exibindo o resultado comparativo
print(df[['Funcionario', 'Salario_k', 'Salario_Log', 'Log_MinMax',
          'Log_Standard', 'Log_Robust']].round(2))
```
