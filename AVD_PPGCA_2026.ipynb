**Normalização**

```python
import pandas as pd

dados = {
    'Candidato': ['Ana', 'Bruno', 'Carlos', 'Daniela', 'Eduardo'],
    'Anos_Experiencia': [1, 3, 5, 8, 10],
    'Pontuacao_Teste': [400, 550, 700, 850, 1000]
}
df = pd.DataFrame(dados)
df.head()
```

```python
# Aplicar as transformações na coluna 'Pontuacao_Teste'
x = df['Pontuacao_Teste']
min_x = x.min()
max_x = x.max()
media_x = x.mean()
desvio_x = x.std()

# 1. Normalização pelo Máximo
df['Teste_Max'] = x / max_x

# 2. Normalização Min-Max ([0, 1])
df['Teste_MinMax'] = (x - min_x) / (max_x - min_x)

# 3. Escalonamento em Intervalo Personalizado [a, b]
a, b = -2, 2
df['Teste_Intervalo'] = a + ((x - min_x) * (b - a)) / (max_x - min_x)

# 4. Padronização Z-Score (Média = 0, Desvio Padrão = 1)
df['Teste_ZScore'] = (x - media_x) / desvio_x


print(df[['Candidato', 'Pontuacao_Teste', 'Teste_Max', 'Teste_MinMax',
          'Teste_Intervalo', 'Teste_ZScore']].round(3))
```

### Gráfico de Radar

```python
# Carregamento do dataset
df_brasil = pd.read_csv('BRAZIL_CITIES.csv', sep=';')

# Cria a coluna 'REGION' no DataFrame, mapeando cada UF (estado) para sua região
mapa_regioes = {
    'AC': 'Norte', 'AM': 'Norte', 'AP': 'Norte', 'PA': 'Norte', 'RO': 'Norte',
    'RR': 'Norte', 'TO': 'Norte', 'AL': 'Nordeste', 'BA': 'Nordeste',
    'CE': 'Nordeste', 'MA': 'Nordeste', 'PB': 'Nordeste', 'PE': 'Nordeste',
    'PI': 'Nordeste', 'RN': 'Nordeste', 'SE': 'Nordeste', 'DF': 'Centro-Oeste',
    'GO': 'Centro-Oeste', 'MS': 'Centro-Oeste', 'MT': 'Centro-Oeste',
    'ES': 'Sudeste', 'MG': 'Sudeste', 'RJ': 'Sudeste', 'SP': 'Sudeste',
    'PR': 'Sul', 'RS': 'Sul', 'SC': 'Sul'
}
df_brasil['REGION'] = df_brasil['STATE'].map(mapa_regioes)

# Seleciona as variáveis que vamos comparar no radar:
variaveis_idhm = ['IDHM_Renda', 'IDHM_Longevidade', 'IDHM_Educacao']
variaveis_gva = ['GVA_AGROPEC', 'GVA_INDUSTRY', 'GVA_SERVICES', 'GVA_PUBLIC']

# Calcula a média de cada variável, agrupando os municípios por região
dados = df_brasil.groupby('REGION')[variaveis_idhm + variaveis_gva].mean()

# O IDHM está em escala 0-1, enquanto o GVA será convertido para %.
# Multiplicar por 100 coloca o IDHM na mesma ordem de grandeza das demais
# variáveis, para nenhum eixo "dominar" o gráfico por causa da escala.
dados[variaveis_idhm] = dados[variaveis_idhm] * 100

# O GVA vem em valores monetários brutos, que variam muito entre regiões
# Por isso, convertemos cada setor para sua % de participação dentro da própria região.
dados['GVA_TOTAL'] = dados[variaveis_gva].sum(axis=1)
for col in variaveis_gva:
    dados[col] = (dados[col] / dados['GVA_TOTAL']) * 100
dados = dados.drop(columns='GVA_TOTAL')  # coluna auxiliar não é mais necessária

# Seleciona apenas 3 regiões para não poluir o gráfico com muitos polígonos
dados = dados.loc[['Sul', 'Nordeste', 'Sudeste']]

# Define os ângulos dos eixos do radar
categorias = list(dados.columns)
n_vars = len(categorias)   # quantidade de eixos
# criar sequência de número igualmente espaçados entre 0 e 2pi
angulos = np.linspace(0, 2*np.pi, n_vars, endpoint=False).tolist()
angulos += angulos[:1]  # repete o primeiro ângulo no final para fechar o polígono

fig, ax = plt.subplots(figsize=(7, 7), subplot_kw=dict(polar=True)) #definir malhar polar
cores = {'Sul': '#5B8FF9', 'Nordeste': '#F98C5C', 'Sudeste': '#5AD8A6'}

# Desenha um polígono por região
for regiao in dados.index:
    valores = dados.loc[regiao].tolist()
    valores += valores[:1]
    ax.plot(angulos, valores, linewidth=2, label=regiao, color=cores[regiao])
    ax.fill(angulos, valores, alpha=0.1, color=cores[regiao])

# Configura os rótulos de cada eixo (nome das variáveis ao redor do círculo)
ax.set_xticks(angulos[:-1])
ax.set_xticklabels(categorias, fontsize=9)
ax.set_yticklabels([])  # oculta os valores numéricos do raio
ax.set_title('Perfil comparativo entre regiões: IDH e composição econômica (%)',
             fontsize=12, pad=20)
ax.legend(loc='best')

plt.tight_layout()
plt.show()
```

```python
# =============================================================
# RV e IQV no exemplo das turmas (k = 5, n = 10)
# =============================================================
turmas = {
    'A': [10, 0, 0, 0, 0],
    'B': [8, 1, 1, 0, 0],
    'C': [4, 3, 3, 0, 0],
    'D': [2, 2, 2, 2, 2]
}
k = 5   # número de categorias possíveis

for nome in turmas:
    f = np.array(turmas[nome])      # frequências de cada classe
    n = f.sum()                     # total de alunos

    rv = 1 - f.max() / n
    iqv = k * (n ** 2 - (f ** 2).sum()) / (n ** 2 * (k - 1))

    print('Turma', nome, '| RV:', round(rv, 2), '| IQV:', round(iqv, 3))
```

```python
# =============================================================
# RV e IQV com dados reais
# =============================================================
for coluna in ['RURAL_URBAN', 'REGIAO']:
    f = df[coluna].value_counts()   # frequências (sem os NaN)
    k = len(f)                      # número de categorias
    n = f.sum()                     # total de valores NÃO ausentes

    rv = 1 - f.max() / n
    iqv = k * (n ** 2 - (f ** 2).sum()) / (n ** 2 * (k - 1))

    print(coluna,
          '| k:', k,
          '| % na moda:', round(f.max() / n * 100, 1),
          '| RV:', round(rv, 2),
          '| IQV:', round(iqv, 2))

# Observação: n = len(df['RURAL_URBAN']) conta também os 3 NaN.
# Com n = f.sum(), os ausentes ficam fora, como no value_counts().
print('len():', len(df['RURAL_URBAN']), '| f.sum():', df['RURAL_URBAN'].value_counts().sum())
```
